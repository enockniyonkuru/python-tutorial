# 2. Python Basics

**In plain English:** A **variable** is a name for a stored value. Every value has a **type**: whole number (`int`), decimal (`float`), text (`str`) or True/False (`bool`). **Operators** (`+`, `-`, `*`, `/`) combine values. A **function** is a reusable recipe: inputs go in, an answer comes out.

**Lab analogy:** A variable is the **label on a tube**. The contents can change but the label stays. A function is a **protocol**: "dilute sample" takes a sample and a dilution factor and gives back the diluted sample. Write it once, use it on every sample.

**Key terms**
- **Variable:** a name for a value, created with `=`, as in `n_cells = 5000`.
- **Function:** a named recipe. You give it **arguments** (inputs) and it **returns** a result.
- **f-string:** text starting with `f` that can hold variables, such as `f"{gene} is ready"`.

**Watch out for**
- `=` **stores** a value (`x = 5`). `==` **compares** two values (`x == 5`).
- `"5"` (text) and `5` (number) are different things and can't be added.
- Python counts from **0**. The first letter of `"BRCA1"` is position 0.
- Writing a function doesn't run it. You must also *call* it: `dilute(100, 10)`.

## Let's run it

**Variables and types.** Four variables, one of each type. You'll see the values, then their types.

````python
n_cells = 5000          # int   (whole number)
viability = 0.92        # float (decimal)
cell_line = "HEK293"    # str   (text)
is_treated = True       # bool  (True / False)

print(n_cells, viability, cell_line, is_treated)
print(type(n_cells), type(viability), type(cell_line), type(is_treated))
````

**Operations.** Everyday lab maths: live cells first, then the main operators. Note that `10 / 3` gives a decimal, `10 // 3` gives `3` and `10 % 3` gives the remainder, `1`.

````python
live_cells = n_cells * viability
print("Live cells:", live_cells)

print(10 + 3)    # add
print(10 - 3)    # subtract
print(10 * 3)    # multiply
print(10 / 3)    # divide
print(10 // 3)   # divide, drop the remainder
print(10 % 3)    # remainder
print(2 ** 10)   # power
````

**Text (strings).** Gene names and sample IDs are text. You'll see `brca1`, `BRCA1_mutant`, a sentence, then `B 1`.

````python
gene = "BRCA1"
print(gene.lower())                          # lowercase
print(gene + "_mutant")                      # join text
print(f"{gene} has {len(gene)} characters")  # f-string
print(gene[0], gene[-1])                     # first and last letter
````

**Lists and appending.** A list holds many values in order, and `.append()` adds one to the end. You'll see a list of four names, then `Number of samples: 4`.

````python
samples = ["ctrl_1", "ctrl_2"]
samples.append("treated_1")
samples.append("treated_2")

print(samples)
print("Number of samples:", len(samples))
````

**Functions.** `dilute` takes two inputs and returns the diluted value. You'll see `10.0` twice. (Defining the function prints nothing, and the output comes from the two calls at the bottom.)

````python
def dilute(concentration, factor):
    """Return the concentration after diluting `factor`-fold."""
    return concentration / factor

print(dilute(100, 10))
print(dilute(50, 5))
````

A function input can have a **default value**, used when you don't provide one. Here `decimals` defaults to 1. You'll see `92.3`, then `92.0`.

````python
def percent(part, total, decimals=1):
    """Return part as a percentage of total."""
    return round(100 * part / total, decimals)

print(percent(923, 1000))      # uses the default, decimals=1
print(percent(923, 1000, 0))   # override the default
````

# 3. Control Flow: Conditionals and Loops

**In plain English:** Code normally runs top to bottom. A **conditional** (`if` / `elif` / `else`) lets it make a decision. A **loop** (`for`, `while`) repeats an action. Python uses **indentation** (4 spaces) to know which lines belong inside the `if` or loop.

**Lab analogy:** A conditional is a **QC checkpoint**: *if* viability is above 90%, go to library prep, *else* repeat the sample. A loop is **pipetting across a 96-well plate**: the same step for A1, A2, A3 and so on.

**Key terms**
- **Condition:** a question with a True/False answer, such as `viability >= 0.90`.
- **Comparison operators:** `>`, `<`, `>=`, `<=`, `==` (equal), `!=` (not equal).
- **Iteration:** one trip around a loop.

**Watch out for**
- Forgetting the `:` at the end of an `if`, `else`, `for` or `while` line.
- Wrong indentation. Lines inside the `if` or loop must be indented.
- A `while` loop that never stops. Click the stop button (or restart the kernel).
- `range(5)` gives `0, 1, 2, 3, 4`, not 1 to 5.

## Let's run it

**`if / elif / else`.** Python checks from the top and runs only the first block that is True. With a viability of 0.85 you'll see `Borderline, proceed with caution`. Try changing the number.

````python
viability = 0.85

if viability >= 0.90:
    print("Great, proceed to library prep")
elif viability >= 0.70:
    print("Borderline, proceed with caution")
else:
    print("Too low, redo the sample")
````

Combine checks with `and` / `or`. A single-cell QC rule usually needs several. You'll see `Cell passes QC`.

````python
n_genes = 2500
mito_percent = 8

if n_genes > 500 and mito_percent < 20:
    print("Cell passes QC")
else:
    print("Cell fails QC")
````

**`for` loops.** Do something for every item in a list. You'll see four "Processing" lines.

````python
samples = ["ctrl_1", "ctrl_2", "treated_1", "treated_2"]

for s in samples:
    print("Processing", s)
````

Use `range(n)` to repeat something a fixed number of times, like counting wells. You'll see `Well 1` to `Well 5`.

````python
for i in range(5):
    print("Well", i + 1)
````

A loop and an `if` together: count the treated samples. You'll see `Treated samples: 2`.

````python
treated_count = 0
for s in samples:
    if "treated" in s:
        treated_count += 1     # same as treated_count = treated_count + 1
print("Treated samples:", treated_count)
````

**`while` loops** repeat for as long as a condition is True. Here, cells double every day until they pass 100,000. You'll see `Took 7 days (now 128000 cells)`.

````python
cells = 1000
days = 0

while cells < 100_000:
    cells *= 2      # cells = cells * 2
    days += 1

print(f"Took {days} days (now {cells} cells)")
````

**List comprehension** is a one-line loop that builds a new list. You'll see it a lot in real code. You'll see `[20, 500, 6, 960, 0]`, then `[10, 250, 480]`.

````python
counts = [10, 250, 3, 480, 0]

doubled = [c * 2 for c in counts]
expressed = [c for c in counts if c > 5]

print(doubled)
print(expressed)
````

# 4. Data Structures: Lists, Sets, Dictionaries

**In plain English:** These are containers that hold many values.
- A **list** `[...]` keeps items **in order** and allows repeats.
- A **set** `{...}` keeps **unique** items, with no order.
- A **dictionary** `{key: value}` is a **lookup table**: give it a key, get back a value.

**Lab analogy:** A list is a **rack of tubes** in order. A set is **the distinct cell types in your sample**: seeing "T cell" a thousand times still counts once. A dictionary is your **sample sheet**: look up "S1" and get "control".

**Key terms**
- **Index:** an item's position. The first is `0`, and `-1` is the last.
- **Slice:** part of a list, such as `genes[1:3]` (start included, end excluded).
- **Key / value:** what you look up / what you get back.

**Watch out for**
- The first item is at position **0**, and a slice `[1:3]` stops *before* 3.
- Sets have no order and no repeats. Use a list if order matters.
- Looking up a missing dictionary key causes a `KeyError`. `.get(key, default)` is the safe way.
- `{}` is an empty *dictionary*, not an empty set.

## Let's run it

**Lists.** Read, slice, add, remove, sort and check membership. You'll see five lines, ending with the sorted list and `True`.

````python
genes = ["TP53", "BRCA1", "MYC", "EGFR"]

print(genes[0])         # first item
print(genes[-1])        # last item
print(genes[1:3])       # items 1 and 2

genes.append("KRAS")    # add to the end
genes.remove("MYC")     # remove by value
genes.sort()            # A to Z

print(genes)
print("TP53" in genes)  # is it in the list?
````

**Sets.** Compare two gene lists from two "labs". You'll see the shared genes (`KRAS`, `MYC`), then all genes, then the genes only in lab A. Set order can look random, which is normal.

````python
lab_a = {"TP53", "MYC", "EGFR", "KRAS"}
lab_b = {"MYC", "BRCA1", "KRAS", "PTEN"}

print("Shared:", lab_a & lab_b)
print("In either:", lab_a | lab_b)
print("Only in A:", lab_a - lab_b)

# A quick way to remove duplicates from a list:
print(set(["T", "B", "T", "NK", "B", "T"]))
````

**Dictionaries.** A sample sheet: look up a value by its key, add an entry, loop over the pairs. You'll see `treated`, the updated dictionary, then one `sample -> condition` line each.

````python
sample_sheet = {"S1": "control", "S2": "treated", "S3": "treated"}

print(sample_sheet["S2"])           # look up
sample_sheet["S4"] = "control"      # add
print(sample_sheet)

for sample, condition in sample_sheet.items():
    print(sample, "->", condition)
````

Dictionary values can be anything, even lists. `.get()` returns a default instead of an error when a key is missing. You'll see `T`, then `unknown`.

````python
experiment = {"name": "PBMC run 1", "cell_types": ["T", "B", "NK"], "n_cells": 5000}

print(experiment["cell_types"][0])
print(experiment.get("species", "unknown"))
````

**Counting with a dictionary** is a tally sheet and a very common pattern. You'll see `{'T': 3, 'B': 2, 'NK': 1}`. (In Part 2, pandas does this for you with `value_counts()`.)

````python
cell_types = ["T", "B", "T", "NK", "B", "T"]

tally = {}
for ct in cell_types:
    tally[ct] = tally.get(ct, 0) + 1    # current count (0 if new) + 1

print(tally)
````

# 5. Object-Oriented Programming Basics

**In plain English:** An **object** bundles data and actions together. A **class** is the blueprint, and each object is one thing built from it. An object's data are its **attributes** (`s1.tissue`) and its actions are its **methods** (`s1.describe()`).

You'll rarely write classes at first, but you'll **use objects all the time**. A pandas table, a scikit-learn model and a Hugging Face model are all objects, and you work with them as `object.method()`.

**Lab analogy:** A class is a **blank sample form** with boxes for ID, tissue and cell count. Each filled-in form is an object. You design the form once and fill in as many as you need.

**Key terms**
- **Class:** the blueprint. **Object:** one thing made from it.
- **Attribute:** data stored on the object. **Method:** a function that belongs to the object.
- **`__init__`:** the setup method that runs when an object is created.
- **`self`:** how a method refers to its own object. Python fills it in for you.

**Watch out for**
- Forgetting `self.` when saving data (`self.tissue = tissue`, not `tissue = tissue`).
- Forgetting the brackets: `s1.describe` is the method itself, `s1.describe()` runs it.
- `__init__` has **two** underscores on each side.

## Let's run it

Define a `Sample` blueprint, then build two objects from it. You'll see two descriptions, then `True False`, then `blood`.

````python
class Sample:
    """A lab sample with an ID, a tissue and a cell count."""

    def __init__(self, sample_id, tissue, n_cells):
        self.sample_id = sample_id      # attributes: data saved on the object
        self.tissue = tissue
        self.n_cells = n_cells

    def describe(self):                 # a method: a function that belongs to the class
        return f"{self.sample_id}: {self.n_cells} cells from {self.tissue}"

    def is_big(self, threshold=3000):
        return self.n_cells >= threshold

s1 = Sample("S1", "blood", 5000)
s2 = Sample("S2", "liver", 1200)

print(s1.describe())
print(s2.describe())
print(s1.is_big(), s2.is_big())
print(s1.tissue)
````

**You've already been using objects.** A list is an object of class `list`, and `.sort()` is one of its methods. You'll see `[1, 2, 3] <class 'list'>`.

````python
my_list = [3, 1, 2]
my_list.sort()
print(my_list, type(my_list))
````

**Optional: inheritance.** A new class can reuse an existing one and add to it. `SingleCellSample` gets everything from `Sample` (including `is_big`) and adds a gene count. You'll see a description ending in `20000 genes detected`, then `True`. Feel free to skip this if you're short on time.

````python
class SingleCellSample(Sample):          # "(Sample)" means: start from Sample
    def __init__(self, sample_id, tissue, n_cells, n_genes):
        super().__init__(sample_id, tissue, n_cells)   # let Sample handle the first three
        self.n_genes = n_genes

    def describe(self):
        return super().describe() + f", {self.n_genes} genes detected"

sc = SingleCellSample("S3", "blood", 8000, 20000)
print(sc.describe())
print(sc.is_big())      # inherited from Sample
````
